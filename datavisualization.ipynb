```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Τα δεδομένα μας
months = ["Ιαν", "Φεβ", "Μαρ", "Απρ", "Μάι", "Ιούν"]
revenues = [12000, 15500, 14200, 19000, 24000, 22500]

# 2. Δημιουργία του γραφήματος (Line Plot)
plt.figure(figsize=(8, 4)) # Ορίζουμε το μέγεθος του παραθύρου (πλάτος, ύψος)
plt.plot(months, revenues, marker='o', color='orange', linewidth=2, linestyle='-')

# 3. Προσθήκη τίτλων και ετικετών (Πάντα απαραίτητα!)
plt.title("Πορεία Εσόδων Πρώτου Εξαμήνου 2026", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Μήνες", fontsize=12)
plt.ylabel("Έσοδα (€)", fontsize=12)

# 4. Προσθήκη πλέγματος (Grid) για να διαβάζεται πιο εύκολα
plt.grid(True, linestyle=':', alpha=0.6)

# 5. Προβολή του γραφήματος
plt.show()
```
