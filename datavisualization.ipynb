```python
import matplotlib.pyplot as plt
import pandas as pd

# 1. Τα δεδομένα μας
months = ["Ιαν", "Φεβ", "Μαρ", "Απρ", "Μάι", "Ιούν"]
revenues = [12000, 15500, 14200, 19000, 24000, 22500]

# 2. Δημιουργία του γραφήματος (Line Plot)
plt.figure(figsize=(8, 4)) # Ορίζουμε το μέγεθος του παραθύρου (πλάτος, ύψος)
plt.plot(months, revenues, marker='o', color='orange', linewidth=2, linestyle='-')

# 3. Προσθήκη τίτλων και ετικετών (Πάντα απαραίτητα!)
plt.title("Πορεία Εσόδων Πρώτου Εξαμήνου 2026", fontsize=14, fontweight='bold', pad=15)
plt.xlabel("Μήνες", fontsize=12)
plt.ylabel("Έσοδα (€)", fontsize=12)

# 4. Προσθήκη πλέγματος (Grid) για να διαβάζεται πιο εύκολα
plt.grid(True, linestyle=':', alpha=0.6)

# 5. Προβολή του γραφήματος
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Δημιουργούμε 1.000 τυχαίους μισθούς με μέσο όρο 1500€ και τυπική απόκλιση 300€
np.random.seed(42)
salaries = np.random.normal(loc=1500, scale=300, size=1000)

# 2. Δημιουργία του Ιστογράμματος
plt.figure(figsize=(8, 5))
# Το bins=20 χωρίζει τα δεδομένα σε 20 "κουτάκια" (μπάρες)
plt.hist(salaries, bins=20, color='crimson', edgecolor='black', alpha=0.7)

# 3. Τίτλοι και Labels
plt.title("Κατανομή Μισθών 1.000 Εργαζομένων", fontsize=14, fontweight='bold')
plt.xlabel("Μισθός (€)", fontsize=12)
plt.ylabel("Πλήθος Εργαζομένων", fontsize=12)

# 4. Εμφάνιση
plt.grid(axis='y', linestyle='--', alpha=0.5) # Grid μόνο στον οριζόντιο άξονα Y
plt.show()
```
