```python
# 1. Το Προσχέδιο (Class)
class User:
    
    # 2. Η μαγική συνάρτηση κατασκευής (__init__)
    # Τρέχει αυτόματα κάθε φορά που φτιάχνουμε έναν νέο χρήστη
    def __init__(self, first_name, last_name,age=None):
        # Αυτά είναι τα Χαρακτηριστικά (Attributes)
        self.first_name = first_name
        self.last_name = last_name
        self.age = age
        self.is_active = True  # Προεπιλογή: Κάθε νέος χρήστης είναι ενεργός
        
    # 3. Μια Συμπεριφορά (Method)
    def introduce(self):
        print(f"Γεια σας, είμαι ο/η {self.first_name} {self.last_name} και είμαι {self.age} χρονων!")
        
    def deactivate_account(self):
        self.is_active = False
        print(f"Ο λογαριασμός του/της {self.first_name} μόλις απενεργοποιήθηκε.")

# ==========================================
# ΠΑΙΖΟΝΤΑΣ ΜΕ ΤΑ ΑΝΤΙΚΕΙΜΕΝΑ (OBJECTS)
# ==========================================

# Φτιάχνουμε δύο διαφορετικά αντικείμενα (χρήστες) από το ίδιο προσχέδιο
user_1 = User("Γιάννης", "Παπαδόπουλος")
user_2 = User("Μαρία", "Γεωργίου")
user_3 = User( 'Dimos', 'Manis', '32')
# Καλούμε τις μεθόδους τους
user_1.introduce()
user_2.introduce()
user_3.introduce()
# Ας κλείσουμε τον λογαριασμό του Γιάννη
user_1.deactivate_account()

# Ελέγχουμε τις ιδιότητές τους (Attributes)
print(f"Ο Γιάννης είναι ενεργός; {user_1.is_active}")
print(f"Η Μαρία είναι ενεργή; {user_2.is_active}")
```

```python
# 1. Φτιάχνουμε το γενικό σχέδιο (Μητρική Κλάση)
class User:
    def __init__(self, first_name, last_name, age):
        self.first_name = first_name
        self.last_name = last_name
        self.age = age
        self.is_active = True
        
    def introduce(self):
        print(f"Γεια σας, είμαι ο/η {self.first_name} {self.last_name}!")

# 2. Φτιάχνουμε ένα ειδικό σχέδιο (Θυγατρική Κλάση)
# Βάζοντας το (User) στην παρένθεση, ο Admin κληρονομεί τα ΠΑΝΤΑ αυτόματα!
class Admin(User):
    
    # Προσθέτουμε μια λειτουργία που την έχει ΜΟΝΟ ο Admin
    def delete_user(self, user_to_delete):
        print(f"Ασφάλεια: Ο Admin {self.first_name} διέγραψε τον χρήστη {user_to_delete.first_name}!")
```

```python
# Δημιουργούμε έναν απλό χρήστη (χρειάζεται: όνομα, επίθετο, ηλικία)
regular_user = User("Γιώργος", "Παπαδόπουλος", 25)

# Δημιουργούμε έναν Admin (κι αυτός χρειάζεται όνομα, επίθετο, ηλικία επειδή είναι User!)
admin_user = Admin("Μαρία", "Γεωργίου", 35)

# ΔΟΚΙΜΗ 1: Ο Admin τρέχει τη μέθοδο introduce() παρόλο που δεν τη γράψαμε μέσα του!
admin_user.introduce()

# ΔΟΚΙΜΗ 2: Ο Admin διαγράφει τον απλό χρήστη
admin_user.delete_user(regular_user)
```

```python
class BankAccount:
    def __init__(self, owner, starting_balance):
        self.owner = owner
        # Το διπλό underscore (__) κάνει το balance PRIVATE (ιδιωτικό)
        self.__balance = starting_balance 
        
    # Μέθοδος για να βλέπουμε το υπόλοιπο (αφού πλέον είναι κρυφό)
    def get_balance(self):
        return self.__balance
        
    # Μέθοδος για κατάθεση (με έλεγχο!)
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"Κατάθεση {amount}€ επιτυχής! Νέο υπόλοιπο: {self.__balance}€")
        else:
            print("Σφάλμα: Το ποσό κατάθεσης πρέπει να είναι θετικό!")

# ==========================================
# ΔΟΚΙΜΗ ΣΤΗΝ ΠΡΑΞΗ
# ==========================================

# Φτιάχνουμε έναν λογαριασμό με 500€
my_account = BankAccount("Γιώργος", 500)

# 1. Προσπαθούμε να δούμε το balance απευθείας:
# print(my_account.__balance) # <- Αν το ξε-σχολιάσεις αυτό, η Python θα σου πετάξει ERROR!

# 2. Ο σωστός τρόπος να δούμε το υπόλοιπο:
print(f"Υπόλοιπο λογαριασμού: {my_account.get_balance()}€")

# 3. Κάνουμε μια κατάθεση
my_account.deposit(150)
```
