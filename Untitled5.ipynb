```python
import pandas as pd

# 1. Δημιουργούμε δεδομένα χρησιμοποιώντας ένα Dictionary της Python
data = {
    "Περιοχή": ["Μαρούσι", "Χαλάνδρι", "Κυψέλη", "Γλυφάδα", "Περιστέρι"],
    "Τιμή (€)": [180000, 240000, 95000, 420000, 130000],
    "Τμ": [85, 100, 70, 120, 75]
}

# 2. Μετατρέπουμε το Dictionary σε Pandas DataFrame (Πίνακα)
df = pd.DataFrame(data)

# 3. Εμφάνιση του πίνακα
print("Ο Πίνακας Δεδομένων μας (DataFrame):")
print(df)
```

```python
# Α. Πώς βλέπουμε ΜΟΝΟ μία στήλη (αυτό είναι ένα Series)
print("Η στήλη με τις Τιμές:")
print(df["Τιμή (€)"])

# Β. Πώς βρίσκουμε τον Μέσο Όρο (Mean) των τιμών
meso_orοs = df["Τιμή (€)"].mean()
print(f"\nΟ μέσος όρος των τιμών είναι: {meso_orοs}€")

# Γ. Φιλτράρισμα: Θέλουμε μόνο τα σπίτια που είναι πάνω από 80 Τετραγωνικά (Τμ)
big_houses = df[df["Τμ"] > 80]
print("\nΣπίτια άνω των 80 Τμ:")
print(big_houses)
```

```python
# Φτιάχνουμε μια custom συνάρτηση για να κατηγοριοποιήσουμε την αποδοτικότητα κάθε ημέρας
df["Αποδοτικότητα"] = df.apply(
    lambda row: "High" if row["Έσοδα (€)"] > 600 and row["Τεμάχια"] > 5 else "Normal", 
    axis=1
)

# Δες πώς φιλτράρουμε γρήγορα χρησιμοποιώντας το .query() (Advanced μέθοδος αντί για το κλασικό φίλτρο)
high_performers = df.query("Αποδοτικότητα == 'High'")

print("--- Ημέρες με Υψηλή Αποδοτικότητα (Query) ---")
print(high_performers.head())
```

```python
import pandas as pd
import numpy as np
```

```python
np.random.seed(42)
n_rows = 200

df_sales = pd.DataFrame({
    "ID_Παραγγελίας": range(1001, 1001 + n_rows),
    "Πόλη": np.random.choice(["Αθήνα", "Θεσσαλονίκη", "Πάτρα", "Ηράκλειο"], size=n_rows),
    "Τμήμα": np.random.choice(["Τεχνολογία", "Μόδα", "Σπίτι", None], size=n_rows, p=[0.4, 0.3, 0.2, 0.1]), # Έχει κενά!
    "Έσοδα": np.random.normal(loc=400, scale=120, size=n_rows).round(2),
    "Επιστροφή": np.random.choice([True, False], size=n_rows, p=[0.15, 0.85])
})

# Εισάγουμε επίτηδες μερικά ακραία λάθη (Outliers & Αρνητικές τιμές)
df_sales.loc[15, "Έσοδα"] = 8500.00  # Ύποπτα τεράστιο ποσό
df_sales.loc[42, "Έσοδα"] = -250.00  # Αδύνατον να υπάρχουν αρνητικά έσοδα

print("--- Το αρχικό dataset δημιουργήθηκε! ---")
print(df_sales.head(10))
```

```python
# =========================================================================
# ΤΑΣΚ 1: DATA CLEANING & ΚΕΝΕΣ ΤΙΜΕΣ (Missing Values)
# =========================================================================
# Α. Βρες πόσες κενές τιμές υπάρχουν στη στήλη "Τμήμα".
# Β. Αντικατάστησε τις κενές τιμές (None/NaN) με τη λέξη "Άγνωστο" χρησιμοποιώντας .fillna()
# Γ. Φιλτράρισε το DataFrame ώστε να πετάξεις εκτός τις γραμμές όπου τα "Έσοδα" είναι αρνητικά (< 0).

# Γράψε τον κώδικα για το Τασκ 1 εδώ:
print("\n[Task 1] Καθαρισμός δεδομένων...")
print(f"Κενά στο Τμήμα πριν: {df_sales.isnull().sum()}")

```

```python
df_sales["Τμήμα"] = df_sales["Τμήμα"].fillna("Άγνωστο")
df_sales = df_sales.query("Έσοδα >= 0") # Κρατάμε μόνο τα θετικά έσοδα
```

```python
#ΤΑΣΚ 2: ADVANCED FEATURE ENGINEERING (Με Lambda & Apply)
# =========================================================================
# Δημιούργησε μια νέα στήλη με όνομα "Κατάσταση_Παραγγελίας".
# Η λογική είναι: 
# - Αν "Επιστροφή" == True -> "Returned"
# - Αν "Επιστροφή" == False ΚΑΙ "Έσοδα" > 500 -> "Target_Achieved"
# - Σε κάθε άλλη περίπτωση -> "Standard"
# (Χρησιμοποίησε .apply() με lambda row: ... και axis=1)

# Γράψε τον κώδικα για το Τασκ 2 εδώ:
print("\n[Task 2] Δημιουργία στήλης Κατάσταση_Παραγγελίας...")

df_sales["Κατάσταση_Παραγγελίας"] = df_sales.apply(
    lambda row: "Returned" if row["Επιστροφή"]==True else ("Target_Achieved" if row["Έσοδα"] > 500 else "Standard"),
    axis=1
)
```

```python
df_sales.head(20)
```

```python
# =========================================================================
# ΤΑΣΚ 3: OUTLIER DETECTION (Στατιστικός Έλεγχος)
# =========================================================================
# Οι Senior Data Scientists βάζουν όρια ασφαλείας. 
# Βρες τις παραγγελίες που τα "Έσοδα" είναι πάνω από 3 τυπικές αποκλίσεις από τον μέσο όρο.
# Όριο = Μέσος Όρος + (3 * Τυπική Απόκλιση)
# Χρησιμοποίησε τις .mean() και .std() των Pandas στη στήλη "Έσοδα".
# Μετά κάνε .query() για να απομονώσεις αυτές τις ακραίες παραγγελίες!

# Γράψε τον κώδικα για το Τασκ 3 εδώ:
print("\n[Task 3] Ανίχνευση Outliers...")
mean_revenue = df_sales["Έσοδα"].mean()
std_revenue = df_sales["Έσοδα"].std()
cutoff = mean_revenue + (3 * std_revenue)

outliers = df_sales.query("Έσοδα > @cutoff") # Το @ σου επιτρέπει να καλέσεις τη μεταβλητή cutoff μέσα στο query!
print(f"Βρέθηκαν {len(outliers)} outliers με έσοδα πάνω από {cutoff:.2f}€:")
print(outliers)
```

```python
# =========================================================================
# ΤΑΣΚ 4: ADVANCED GROUPBY & REPORTING
# =========================================================================
# Τώρα που τα δεδομένα είναι καθαρά, φτιάξε το τελικό report για τη διοίκηση.
# Κάνε .groupby() ανά "Πόλη" και "Τμήμα" και υπολόγισε με την .agg():
# 1. Το σύνολο των εσόδων
# 2. Τη μέση τιμή των εσόδων
# 3. Το πλήθος των παραγγελιών (χρησιμοποίησε την πράξη "count")
# Μην ξεχάσεις το .reset_index() στο τέλος!

# Γράψε τον κώδικα για το Τασκ 4 εδώ:
print("\n[Task 4] Τελικό Συγκεντρωτικό Report...")
final_report = df_sales.groupby(["Πόλη", "Τμήμα"]).agg(
    Συνολικά_Έσοδα=("Έσοδα", "sum"),
    Μέσα_Έσοδα=("Έσοδα", "mean"),
    Πλήθος_Παραγγελιών=("ID_Παραγγελίας", "count")
).reset_index()

print(final_report.head(15))
```
